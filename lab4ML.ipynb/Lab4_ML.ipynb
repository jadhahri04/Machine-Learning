```python
import numpy as np

cols_to_scale = ['previous_exam_score', 'previous_gpa', 'attendance_percentage',
                 'study_hours_per_day', 'exam_anxiety_level', 'exam_score']

df_standardized = df[cols_to_scale].copy()
df_standardized = (df_standardized - df_standardized.mean()) / df_standardized.std(ddof=0)

df_standardized.head()
```
