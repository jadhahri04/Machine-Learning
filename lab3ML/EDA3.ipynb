```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
```

```python
df = pd.read_csv("recent_movies_2024_2026.csv")
df.head()
```

```python
df.dtypes
```

```python
print("Missing values per column:\n")
print(df.isna().sum())
```

```python
print("Check Duplicates:\n")
df.duplicated()[df.duplicated()==True]
```

```python
print(" Number of rows and columns:\n")

print("Shape (rows, columns): ", df.shape,"\n")

print("number of rows: ", df.shape[0])
print("number of columns: ", df.shape[1])
```

```python
print("Summary statistics for numerical columns:\n")
df.describe(include='all')
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['rating_out_of_10'], bins=20, kde=True, color='steelblue')
plt.title('Distribution of Movie Ratings')
plt.xlabel('Rating (out of 10)')
plt.ylabel('Frequency')
plt.show()
```

```python
from collections import Counter
all_genres = df['genres'].dropna().str.split(',').explode().str.strip()
genre_counts = all_genres.value_counts().head(15)

plt.figure(figsize=(10, 6))
genre_counts.plot(kind='bar')
plt.title("Top 15 Most Common Genres")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df['content_rating'].value_counts().plot(kind='bar')
plt.title("Distribution of Content Ratings")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='runtime_minutes', y='release_month', data=df)
plt.title('Runtime Minutes by Release Month')
plt.xlabel('Runtime (minutes)')
plt.ylabel('Release Month')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='budget_usd', y='vote_count', data=df)
plt.title('Budget vs Vote Count')
plt.xlabel('Budget (USD)')
plt.ylabel('Vote Count')
plt.show()
```

```python
numeric_cols = ['rating_out_of_10', 'vote_count', 'runtime_minutes', 'box_office_usd', 'budget_usd']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Between Numeric Features')
plt.show()
```
